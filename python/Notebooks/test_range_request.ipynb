# This cell is just a test that we are able to get anonymous access and that range requests are configured correctly.

```python
import s3fs

fs = s3fs.S3FileSystem(
    anon=True,
    client_kwargs={
        "endpoint_url": "https://projects.pawsey.org.au"
    },
)

# read a slice (forces range requests)
with fs.open("zarr-data-stream-test/output/SWWA/WA-DWER/ERA5/historical/r1i1p1f1/R3/v1/mon/pr/pr_SWWA_ERA5_historical_r1i1p1f1_R3_v1_mon_201001-201012.nc", "rb") as f:
    f.seek(1000)
    print(f.read(100))
```

```python
import boto3
from botocore import UNSIGNED
from botocore.config import Config

s3 = boto3.client(
    "s3",
    endpoint_url="https://projects.pawsey.org.au",
    config=Config(signature_version=UNSIGNED),
)

resp = s3.get_object(
    Bucket="zarr-data-stream-test",
    Key="output/SWWA/WA-DWER/ERA5/historical/r1i1p1f1/R3/v1/mon/pr/pr_SWWA_ERA5_historical_r1i1p1f1_R3_v1_mon_201001-201012.nc",
    Range="bytes=1000-1099",
)

print("Status code:", resp["ResponseMetadata"]["HTTPStatusCode"])
print("Content-Range:", resp["ResponseMetadata"]["HTTPHeaders"].get("content-range"))
```
